## 1. La aplicación con dos retrievers

Sin modelo. El índice de L8 en una celda y los dos retrievers que comparamos: **coseno** (los 3 fragmentos más parecidos según los embeddings) y **RRF**, que fusiona el ranking por coseno con el de BM25 (coincidencia de palabras), como en L9. Los dos devuelven 3 fragmentos.



<details><summary>Código Mermaid</summary>

```mermaid
flowchart LR
    Q(["Pregunta"]) --> C["coseno<br/>embeddings"]
    Q --> B["BM25<br/>palabras"]
    C --> R1["retrieve_cosine<br/>top 3"]
    C --> F["RRF<br/>fusiona rankings"]
    B --> F
    F --> R2["retrieve_rrf<br/>top 3"]
```

</details>

````python
import re
from pathlib import Path

from langchain_core.documents import Document
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langsmith import traceable
from pypdf import PdfReader
from rank_bm25 import BM25Plus

pages = []
for path in sorted(Path("../datasets").glob("acme_*.pdf")):
    for number, page in enumerate(PdfReader(path).pages, 1):
        pages.append(Document(page.extract_text(), metadata={"source": path.name, "page": number}))

chunks = RecursiveCharacterTextSplitter(chunk_size=400, chunk_overlap=60).split_documents(pages)
vector_store = InMemoryVectorStore.from_documents(chunks, embeddings)


def tokens(text):
    return re.findall(r"\w+", text.lower())


bm25 = BM25Plus([tokens(chunk.page_content) for chunk in chunks])
chunk_index = {chunk.page_content: i for i, chunk in enumerate(chunks)}


@traceable(run_type="retriever", name="recuperar · coseno")
def retrieve_cosine(question):
    return vector_store.similarity_search(question, k=3)


@traceable(run_type="retriever", name="recuperar · RRF")
def retrieve_rrf(question, k=60):
    cosine = [chunk_index[doc.page_content] for doc in vector_store.similarity_search(question, k=len(chunks))]
    word_scores = bm25.get_scores(tokens(question))
    words = sorted(range(len(chunks)), key=lambda i: -word_scores[i])
    scores = {}
    for ranking in (cosine, words):
        for position, i in enumerate(ranking, 1):
            scores[i] = scores.get(i, 0) + 1 / (k + position)
    return [chunks[i] for i in sorted(scores, key=scores.get, reverse=True)[:3]]


print(len(chunks), "fragmentos")
````

La aplicación que evaluamos: recibe la pregunta, recupera con el retriever que le pasemos y responde con el prompt de L8. Devuelve, además de la respuesta, **el contexto** (lo necesita el juez de sustento), **las fuentes** (las necesitan los evaluadores de código) y **los tokens** (para el costo). `cosine_app` y `rrf_app` son la misma aplicación con un retriever distinto: son las dos configuraciones del experimento.

`@traceable` hace que cada paso aparezca en la traza de LangSmith. Sin él, `evaluate` registra la aplicación como una caja negra llamada `Target` con la llamada a Claude adentro, y no se ve qué fragmentos recuperó cada retriever. Con `run_type="retriever"`, LangSmith muestra los fragmentos como documentos. Fuera de `evaluate` el tracing está apagado y el decorador no hace nada.



<details><summary>Código Mermaid</summary>

```mermaid
flowchart LR
    A["RAG · coseno<br/>pregunta → respuesta, contexto, fuentes"] --> R["recuperar · coseno<br/>3 fragmentos"]
    A --> C["ChatAnthropic<br/>prompt con los fragmentos → respuesta"]
```

</details>

````python
from langchain_core.prompts import ChatPromptTemplate

SYSTEM_PROMPT = """\
Eres el asistente de soporte de ACME, una fintech.
Responde solo con la información de los fragmentos de documentos que recibes.
Después de cada dato, cita su fuente con el formato [archivo, p. N], tal como aparece en el fragmento.
Si los fragmentos no contienen la respuesta, o solo una parte, di qué no está en los documentos consultados y no lo inventes.
No prometas aprobaciones ni des asesoría financiera personalizada.
"""

prompt = ChatPromptTemplate.from_messages(
    [("system", SYSTEM_PROMPT), ("human", "Fragmentos:\n\n{context}\n\nPregunta: {question}")]
)


def rag_answer(question, retrieve):
    docs = retrieve(question)
    context = "\n\n".join(f"[{doc.metadata['source']}, p. {doc.metadata['page']}]\n{doc.page_content}" for doc in docs)
    answer = model.invoke(prompt.format_messages(context=context, question=question))
    return {
        "answer": answer.text,
        "context": context,
        "sources": [[doc.metadata["source"], doc.metadata["page"]] for doc in docs],
        "input_tokens": answer.usage_metadata["input_tokens"],
        "output_tokens": answer.usage_metadata["output_tokens"],
    }


@traceable(name="RAG · coseno")
def cosine_app(inputs):
    return rag_answer(inputs["question"], retrieve_cosine)


@traceable(name="RAG · RRF")
def rrf_app(inputs):
    return rag_answer(inputs["question"], retrieve_rrf)
````

## 3. Evaluadores de código

Sin modelo. Un **evaluador** recibe la entrada, la salida de la aplicación y la referencia, y devuelve una puntuación con un nombre (`key`). Los de código son baratos, rápidos y siempre dan el mismo resultado:

- `retrieval_hit`: ¿la página esperada está entre los 3 fragmentos? Es el acierto en el top 3 de L9, ahora por ejemplo. En las preguntas sin respuesta no aplica (`None`).
- `citations_ok`: ¿todas las citas `[archivo, p. N]` vienen de los fragmentos recuperados, como en la sección 6 de L8? Si la pregunta tiene respuesta, además exige al menos una cita.

````python
def retrieval_hit(outputs, reference_outputs):
    if reference_outputs["source"] is None:
        return {"key": "retrieval_hit", "score": None, "comment": "Pregunta sin respuesta en los documentos."}
    expected = [reference_outputs["source"], reference_outputs["page"]]
    return {"key": "retrieval_hit", "score": expected in outputs["sources"]}


def citations_ok(outputs, reference_outputs):
    cited = {(source, int(page)) for source, page in re.findall(r"\[([\w.-]+\.pdf), p\. (\d+)\]", outputs["answer"])}
    retrieved = {(source, page) for source, page in outputs["sources"]}
    needs_citation = reference_outputs["source"] is not None
    return {"key": "citations_ok", "score": cited <= retrieved and (bool(cited) or not needs_citation)}
````

````python
fake_output = {"answer": "Hasta 3.000.000 COP [acme_credit_policy.pdf, p. 1].", "sources": [["acme_credit_policy.pdf", 1]]}
fake_reference = {"source": "acme_credit_policy.pdf", "page": 1}

print(retrieval_hit(fake_output, fake_reference))
print(citations_ok(fake_output, fake_reference))
print(citations_ok({**fake_output, "answer": "Hasta 3.000.000 COP [acme_credit_policy.pdf, p. 2]."}, fake_reference))
````

## 4. El juez con las rúbricas de OpenEvals

Que la página llegue y que las citas existan no dice si la respuesta es **correcta** ni si **cada afirmación** sale de los fragmentos. Para eso usamos un **juez**: un modelo que lee la respuesta con una rúbrica y da una puntuación con su razonamiento. Es la técnica *LLM-as-a-judge*.

OpenEvals trae rúbricas probadas para esto, como `CORRECTNESS_PROMPT` y `RAG_GROUNDEDNESS_PROMPT`. Abajo están traducidas y ajustadas a nuestras preguntas, por ejemplo para aceptar «no está en los documentos» cuando esa es la referencia. Imprimimos el comienzo de la original para compararla.



<details><summary>Código Mermaid</summary>

```mermaid
flowchart LR
    I["pregunta"] --> CJ["juez de corrección"]
    O["respuesta"] --> CJ
    R["referencia"] --> CJ
    O --> GJ["juez de sustento"]
    X["fragmentos"] --> GJ
    CJ --> S1(["correctness<br/>score + comment"])
    GJ --> S2(["groundedness<br/>score + comment"])
```

</details>

El juez es Claude Haiku con **salida estructurada**: devuelve un `Judgement` con `reasoning` y `score`. Dos detalles importan:

- `method="json_schema"` usa las salidas estructuradas nativas de Claude, que **garantizan** el esquema. OpenEvals ofrece `create_llm_as_judge`, que arma el juez en una línea, pero pide la salida con el método por defecto (una herramienta, sin garantía), y algunos juicios pueden llegar sin `score`.
- `max_tokens=2000`: el juez escribe su razonamiento antes de puntuar. Con los 500 de las respuestas, muchos juicios de sustento se cortan.

Usar el mismo modelo que responde es barato, pero tiene un riesgo: un modelo tiende a aprobar respuestas que se parecen a las suyas. Por eso la sección 5 lo contrasta con personas.

````python
from langchain_core.messages import HumanMessage
from openevals.prompts import CORRECTNESS_PROMPT as OPENEVALS_CORRECTNESS_PROMPT
from pydantic import BaseModel, Field

print(OPENEVALS_CORRECTNESS_PROMPT[:300], "...")

CORRECTNESS_PROMPT = """\
Eres un revisor experto que evalúa si la respuesta de un asistente es correcta.

<rubrica>
Una respuesta correcta:
- Da la misma información que la respuesta de referencia, aunque use otras palabras.
- No contiene errores en cifras, plazos ni nombres.
- Responde todas las partes de la pregunta.
- Si la referencia dice que el dato no está en los documentos, la respuesta también lo reconoce y no inventa uno.

Penaliza:
- Cifras, plazos o condiciones distintas a las de la referencia.
- Respuestas incompletas que omiten información clave de la referencia.
- Datos inventados que la referencia no contiene.
</rubrica>

<instrucciones>
- Compara la respuesta con la referencia; no juzgues el estilo ni la longitud.
- Las citas [archivo, p. N] no hacen correcta ni incorrecta una respuesta.
- Explica tu razonamiento en español antes de dar la puntuación.
</instrucciones>

<pregunta>
{inputs}
</pregunta>

<respuesta>
{outputs}
</respuesta>

<referencia>
{reference_outputs}
</referencia>
"""

GROUNDEDNESS_PROMPT = """\
Eres un revisor experto que evalúa si la respuesta de un asistente está sustentada en los fragmentos recuperados.

<rubrica>
Una respuesta sustentada:
- Solo afirma lo que dicen los fragmentos, o lo que se deduce directamente de ellos.
- No añade datos externos, cifras ni condiciones que no estén en los fragmentos.
- Si los fragmentos no contienen la respuesta, lo dice en lugar de completarla.

Una respuesta no sustentada:
- Afirma algo que ningún fragmento dice o lo contradice.
- Mezcla información de forma que cambia su sentido.
</rubrica>

<instrucciones>
- Revisa cada afirmación de la respuesta y busca el fragmento que la respalda.
- Juzga solo contra los fragmentos, aunque sepas otra cosa del tema.
- Explica tu razonamiento en español antes de dar la puntuación.
</instrucciones>

<fragmentos>
{context}
</fragmentos>

<respuesta>
{outputs}
</respuesta>
"""


class Judgement(BaseModel):
    """Puntuación del juez con su razonamiento."""

    reasoning: str = Field(description="Razonamiento en español que termina con la puntuación asignada.")
    score: bool = Field(description="true si la respuesta cumple la rúbrica; false si no.")


judge_model = model.model_copy(update={"max_tokens": 2000})
judge = judge_model.with_structured_output(Judgement, method="json_schema")
````

Con OpenEvals el juez sería una línea por rúbrica. Funciona, pero sin la garantía del esquema:

```python
from openevals.llm import create_llm_as_judge

correctness_judge = create_llm_as_judge(prompt=CORRECTNESS_PROMPT, judge=judge_model, feedback_key="correctness")
```

Si el juez no devuelve el formato esperado, el error no debe confundirse con una respuesta mala. Aquí, si el juez falla, devolvemos `score=None` y un comentario que empieza por «Formato inválido», para contarlos aparte en la sección 10. Los dos evaluadores usan la misma función `run_judge`; solo cambian la rúbrica y lo que se le pasa.

````python
def run_judge(key, prompt):
    try:
        judgement = judge.invoke([HumanMessage(prompt)])
        return {"key": key, "score": judgement.score, "comment": judgement.reasoning}
    except Exception as error:  # noqa: BLE001 -- el juez no devolvió el formato esperado.
        return {"key": key, "score": None, "comment": f"Formato inválido: {error}"}


def correctness(inputs, outputs, reference_outputs):
    prompt = CORRECTNESS_PROMPT.format(
        inputs=inputs["question"], outputs=outputs["answer"], reference_outputs=reference_outputs["reference"]
    )
    return run_judge("correctness", prompt)


def groundedness(outputs):
    return run_judge("groundedness", GROUNDEDNESS_PROMPT.format(context=outputs["context"], outputs=outputs["answer"]))
````

## 5. Calibrar el juez con revisión humana

Antes de usar el juez para decidir, comprobamos si coincide con una persona. `datasets/acme_review_set.csv` tiene **8 respuestas fijas** a preguntas del dataset, escritas a propósito: algunas correctas (una sin cita, una en paráfrasis) y otras con errores típicos (un puntaje del otro producto, un plazo cambiado, una respuesta incompleta, una tasa inventada). Una persona las revisó con la referencia y marcó `human_correct`.

Son fijas para que la comparación no cambie con cada ejecución. En un proyecto real se revisan respuestas de la propia aplicación, por ejemplo con las *annotation queues* de LangSmith (**8 llamadas**).

`tracing_context(enabled=True, project_name=PROJECT)` activa las trazas solo para este bloque y las manda al proyecto del notebook. Dentro, `trace` agrupa las 8 llamadas del juez en **una sola traza**, «L11 · calibración», con tu nombre como etiqueta. Sin `trace`, cada juicio aparecería como una fila suelta en el proyecto.



<details><summary>Código Mermaid</summary>

```mermaid
flowchart LR
    RS["acme_review_set.csv<br/>respuesta + human_correct"] --> J["juez de corrección"]
    J --> C{"¿juez = persona?"}
    C --> A(["acuerdo<br/>y desacuerdos"])
```

</details>

````python
from langsmith import trace, tracing_context

review = pd.read_csv("../datasets/acme_review_set.csv").merge(questions[["id", "question", "reference"]], on="id")

with tracing_context(enabled=True, project_name=PROJECT), trace(
    "L11 · calibración", tags=[PARTICIPANT], metadata={"participant": PARTICIPANT}, client=client
) as calibration_run:
    judgements = [
        correctness({"question": row.question}, {"answer": row.answer}, {"reference": row.reference})
        for row in review.itertuples()
    ]
review["judge_correct"] = [judgement["score"] for judgement in judgements]
review["judge_comment"] = [judgement["comment"] for judgement in judgements]

agreement = (review.judge_correct == review.human_correct.astype(bool)).mean()
print(f"Acuerdo juez-persona: {agreement:.0%}")
review[["review_id", "id", "answer", "human_correct", "judge_correct"]]
````

¿En cuáles no coinciden? Lee el comentario del juez en los desacuerdos:

````python
for row in review[review.judge_correct != review.human_correct.astype(bool)].itertuples():
    print(f"{row.review_id} · persona: {row.note}")
    print(f"     juez: {row.judge_comment[:300]}\n")
````

Con 8 ejemplos, cada desacuerdo mueve el acuerdo 12,5 puntos. Si el juez falla en un tipo de error (por ejemplo, aprueba respuestas incompletas), se ajusta la rúbrica y se vuelve a calibrar **antes** de los experimentos. Un acuerdo alto en 8 ejemplos no garantiza que el juez sea bueno: con más ejemplos revisados la estimación es más confiable.

## 6. Dos experimentos

`evaluate` ejecuta la aplicación con cada ejemplo, aplica los evaluadores y registra todo en LangSmith como un **experimento**. Lo llamamos dos veces, una por configuración, con los mismos ejemplos y los mismos evaluadores (**15 llamadas por experimento**: 5 respuestas y 10 juicios).

## 8. Qué hay dentro de un experimento

Sin modelo. En LangSmith cada experimento es un **proyecto propio**: `l11-<nombre>-coseno` guarda las trazas de la rama coseno y `l11-<nombre>-rrf`, las de RRF. La calibración va aparte, en `advancedgenai-l11`.



<details><summary>Código Mermaid</summary>

```mermaid
flowchart TB
    subgraph X1["Experimento l11-nombre-coseno · 5 trazas"]
        T1["Traza RAG · coseno<br/>inputs: pregunta · outputs: respuesta, contexto, fuentes"]
        T1 --> S1["recuperar · coseno<br/>retriever: 3 fragmentos"]
        T1 --> S2["ChatAnthropic<br/>llm: prompt → respuesta, tokens"]
        T1 -.-> F1["feedback<br/>retrieval_hit · citations_ok · correctness · groundedness"]
    end
    E["Ejemplo del dataset<br/>pregunta + referencia"] -.-> T1
```

</details>

- **Traza**: una ejecución completa de la aplicación para un ejemplo. Es lo que cuenta contra el límite mensual: 5 por experimento, más la de calibración, son **11**.
- **Paso** (*run*): cada parte dentro de la traza, con su tipo (`retriever`, `llm`), entradas, salidas y tiempos. Añadir pasos con `@traceable` no suma trazas.
- **Ejemplo**: la fila del dataset con la que se ejecutó, con la referencia que usan los evaluadores.
- **Feedback**: la puntuación de cada evaluador, pegada a la traza, con el razonamiento del juez como comentario.

`evaluate` devuelve en el notebook los mismos objetos que sube a LangSmith: por cada ejemplo, `row["run"]` (la traza con sus pasos), `row["example"]` y `row["evaluation_results"]`. No hace falta volver a pedirlos a la API. Abramos una traza de cada rama, la misma pregunta:

````python
def seconds(run):
    return (run.end_time - run.start_time).total_seconds()


def show_trace(results, question_id):
    row = next(row for row in results if row["example"].metadata["id"] == question_id)
    run, example = row["run"], row["example"]
    print(f"Traza «{run.name}» · {seconds(run):.2f} s")
    print("  pregunta:  ", example.inputs["question"])
    print("  referencia:", example.outputs["reference"])
    for step in run.child_runs:
        print(f"  └ paso «{step.name}» ({step.run_type}) · {seconds(step):.2f} s")
        if step.run_type == "retriever":
            for doc in step.outputs["output"]:
                print(f"      [{doc.metadata['source']}, p. {doc.metadata['page']}] {doc.page_content[:70]!r}…")
    print("  respuesta: ", run.outputs["answer"])
    print(f"  tokens:     {run.outputs['input_tokens']} de entrada, {run.outputs['output_tokens']} de salida")
    for result in row["evaluation_results"]["results"]:
        print(f"  feedback {result.key}: {result.score}", f"· {result.comment[:120]}…" if result.comment else "")
    print()


show_trace(cosine_results, "q07")
show_trace(rrf_results, "q07")
````

Es lo mismo que se ve al abrir la traza en LangSmith: el árbol de pasos a la izquierda, entradas y salidas de cada paso y, en la pestaña **Feedback**, las puntuaciones. Fíjate en qué fragmentos trajo cada retriever para «otro banco» y en cuánto tiempo se va en recuperar frente a esperar a Claude.

## 9. Resumen por configuración y por pregunta

Sin modelo. `to_pandas()` trae una fila por ejemplo con las salidas, las puntuaciones (`feedback.*`) y el tiempo de ejecución. Resumimos cada experimento: el promedio de cada evaluador, la latencia media de la aplicación y el costo de las respuestas con los precios de Claude Haiku 5.5 (US$0,10 por millón de tokens de entrada y US$0,50 por millón de salida). El costo de los juicios no está incluido.

````python
PRICE_PER_MILLION = {"input": 0.10, "output": 0.50}
METRICS = ["retrieval_hit", "citations_ok", "correctness", "groundedness"]
cosine_table, rrf_table = cosine_results.to_pandas(), rrf_results.to_pandas()


def summarize(table):
    cost = table["outputs.input_tokens"].sum() * PRICE_PER_MILLION["input"] + table["outputs.output_tokens"].sum() * PRICE_PER_MILLION["output"]
    return {
        **{metric: table[f"feedback.{metric}"].mean() for metric in METRICS},
        "latencia_s": table["execution_time"].mean(),
        "costo_usd": cost / 1e6,
    }


summary = pd.DataFrame({"coseno": summarize(cosine_table), "rrf": summarize(rrf_table)}).round(4)
summary
````

Un promedio de 5 preguntas esconde mucho. La vista **Compare** de LangSmith pone cada pregunta en una fila con sus puntuaciones; aquí, lo mismo (1 = pasa, 0 = no pasa, vacío = sin puntuación):

````python
by_question = pd.concat(
    {name: table.set_index("inputs.question")[[f"feedback.{metric}" for metric in METRICS]].astype(float)
     for name, table in {"coseno": cosine_table, "rrf": rrf_table}.items()},
    axis=1,
).rename(columns=lambda column: column.removeprefix("feedback."), level=1)
by_question
````

Y lo que respondió cada configuración, para leerlo junto a las puntuaciones:

````python
answers = cosine_table[["inputs.question", "outputs.answer"]].merge(
    rrf_table[["inputs.question", "outputs.answer"]], on="inputs.question", suffixes=("_cosine", "_rrf")
)
for row in answers.itertuples(index=False):
    print(row[0])
    print("  coseno:", row[1])
    print("  RRF:   ", row[2], "\n")
````

En nuestra ejecución, RRF recuperó la página correcta en las cuatro preguntas con respuesta y coseno falló en q07 («otro banco»), igual que en L9: con la página equivocada, la respuesta de coseno dijo que no podía confirmarlo. ¿Cuál elegirían para el asistente de ACME? Con 5 preguntas, una sola respuesta cambia la corrección en 20 puntos: es una señal para mirar los casos, no una prueba. Para decidir hacen falta más preguntas y repetir los experimentos, porque las respuestas de Claude varían entre ejecuciones (`num_repetitions` en `evaluate`).

## 10. Fallos de formato del juez

Sin modelo. Contamos los juicios que terminaron en «Formato inválido» en los dos experimentos. Si los hay, esos ejemplos no tienen puntuación: no se cuentan como respuestas malas ni buenas, y conviene revisarlos a mano.

````python
format_failures = {
    name: sum(
        result.comment.startswith("Formato inválido")
        for row in results
        for result in row["evaluation_results"]["results"]
        if result.key in {"correctness", "groundedness"} and result.comment
    )
    for name, results in {"coseno": cosine_results, "rrf": rrf_results}.items()
}
format_failures
````
