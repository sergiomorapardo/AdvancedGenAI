## 2 · El grafo del agente

`create_agent` arma el bucle clásico de ReAct: el nodo `model` decide si contesta o si pide una
herramienta; si la pide, pasa por `tools` y vuelve a `model`. Las flechas punteadas son las
aristas condicionales, es decir, las que dependen de esa decisión.

```python
# draw_mermaid_png() dibuja el diagrama con el servicio mermaid.ink: necesita conexión a internet.
display(Image(agent.get_graph().draw_mermaid_png()))
```
