```python
from dotenv import find_dotenv, load_dotenv

from langchain_anthropic import ChatAnthropic
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
```

```python
load_dotenv(find_dotenv())
```

```python
haiku_4_5 = ChatAnthropic(
    model="claude-haiku-4-5",
    temperature=0,
    max_tokens=400,  # 150 cortaba las respuestas a media frase
)
```

```python
response = haiku_4_5.invoke("Hola ¿comó estas?")
response.pretty_print()
```

```python
msg_1 = SystemMessage(content="You're a helpful assistant thaht talks in spanish")
msg_2 = HumanMessage(content="Me llamo Pepito")
msg_3 = AIMessage(content="Hola Pepito, ¿cómo estás?")
msg_4 = HumanMessage(content="¿cómo me llamo?")

history = [
    msg_1,
    msg_2,
    msg_3,
    msg_4,
]

history
```

```python
response = haiku_4_5.invoke(history)
response.pretty_print()
```

```python
from langchain.chat_models import init_chat_model

haiku = init_chat_model("anthropic:claude-haiku-4-5", temperature=0)
response = haiku.invoke(history)

response.pretty_print()
```
